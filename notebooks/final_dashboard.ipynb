## 2. Environment Design

A custom Gymnasium environment (`PricingEnv`) was built to simulate the 
dynamic pricing problem for travel and hospitality bookings. The state 
space consists of two variables — remaining inventory (0 to 100 units) 
and days until departure (0 to 30 days) — while the action space offers 
10 discrete price level bins ranging from lowest to highest. At each 
step, a stochastic demand function using a logistic curve determines 
how many units are sold, with purchase probability decreasing as price 
increases and increasing as the deadline approaches to reflect 
last-minute booking urgency. The reward at each step is calculated as 
price level multiplied by units sold, and an episode terminates when 
either the inventory is fully sold or the days until departure reach 
zero. The environment was validated in Week 1 through 100 random-action 
episodes, confirming stable behavior with zero crashes or invalid states.

**Demand Model:**
A stochastic demand function using a logistic curve determines how many 
units sell at each step. Purchase probability:
- Decreases as price increases
- Increases as the deadline approaches (urgency effect)

**Episode Termination:**
An episode ends when either:
- `remaining_inventory` reaches 0 (sold out), or
- `days_until_departure` reaches 0 (season ended)

**Validation:**
The environment was stress-tested with 100 random-action episodes in 
Week 1 to confirm stability (0 crashes, no invalid states).

## 4. Policy Analysis

### Observed Pricing Behavior

Analysis of agent trajectories across simulated episodes reveals the 
following patterns:

- **Time-Based Discount Agent**: Prices decay predictably by ~10% per 
  day, resulting in aggressive last-minute discounting regardless of 
  actual demand or remaining inventory.
- **Demand-Based Agent**: Prices adjust dynamically based on the 
  inventory-to-time ratio — raising prices when inventory is selling 
  fast relative to time remaining, and lowering prices when inventory 
  is high relative to time left, better reflecting real demand signals 
  than a fixed decay schedule.
- **Learned Agents (Q-Learning / DQN)**: Expected to discover pricing 
  policies that combine both effects — discounting near the deadline 
  *only when* inventory-clearing risk is high, rather than 
  unconditionally, since the reward signal directly penalizes 
  underpricing when demand is otherwise strong.

### Edge Cases to Monitor
Learned policies carry a risk of discovering degenerate behavior, such 
as pricing near zero purely to guarantee a sale late in the season. 
This motivates the safety bounds outlined in Section 5.

## 5. Business Recommendations

### Recommended Safety Bounds
To prevent the pricing agent from behaving detrimentally (e.g., pricing 
near zero purely to force a sale), a Revenue Manager should configure:
- **Minimum price bound**: prevents excessive discounting that erodes margin
- **Maximum price bound**: prevents pricing so high that sell-through fails entirely

### Deployment Readiness
- The environment and evaluation framework have been validated across 
  1,000 simulated seasons per agent, with statistically significant 
  performance differences confirmed via paired t-testing
- Adaptive strategies (time-based, demand-based, and learned policies) 
  outperform static pricing in both mean revenue and sell-through rate

### Monitoring Guidance
Once deployed, the pricing agent should be monitored for:
- Sudden drops in sell-through rate (may indicate pricing too high)
- Excessive last-minute discounting (may indicate reward function 
  encouraging inventory dumping)
- Periodic re-evaluation against updated heuristic baselines as market 
  conditions change
