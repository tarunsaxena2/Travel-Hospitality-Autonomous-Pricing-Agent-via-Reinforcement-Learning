## 1. Project Summary

### Goal
Train a Deep Q-Network (DQN) agent to autonomously set dynamic prices 
for travel/hospitality bookings, outperforming heuristic baselines 
(fixed, time-based discount, demand-based) in mean episodic revenue 
across 1,000 simulated seasons.

### Tech Stack
Python, Gymnasium, PyTorch, NumPy, Pandas, Matplotlib, Seaborn

### MDP Formulation
- **State**: [remaining_inventory, days_until_departure]
- **Action**: Discrete price level bins (10 levels)
- **Reward**: price_level × units_sold
- **Horizon**: Episode ends when inventory = 0 or days_until_departure = 0
