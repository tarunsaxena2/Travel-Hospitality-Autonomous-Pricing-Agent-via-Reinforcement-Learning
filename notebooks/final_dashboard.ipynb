## 2. Environment Design

A custom Gymnasium environment (`PricingEnv`) was built to simulate the 
dynamic pricing problem for travel and hospitality bookings. The state 
space consists of two variables — remaining inventory (0 to 100 units) 
and days until departure (0 to 30 days) — while the action space offers 
10 discrete price level bins ranging from lowest to highest. At each 
step, a stochastic demand function using a logistic curve determines 
how many units are sold, with purchase probability decreasing as price 
increases and increasing as the deadline approaches to reflect 
last-minute booking urgency. The reward at each step is calculated as 
price level multiplied by units sold, and an episode terminates when 
either the inventory is fully sold or the days until departure reach 
zero. The environment was validated in Week 1 through 100 random-action 
episodes, confirming stable behavior with zero crashes or invalid states.

**Demand Model:**
A stochastic demand function using a logistic curve determines how many 
units sell at each step. Purchase probability:
- Decreases as price increases
- Increases as the deadline approaches (urgency effect)

**Episode Termination:**
An episode ends when either:
- `remaining_inventory` reaches 0 (sold out), or
- `days_until_departure` reaches 0 (season ended)

**Validation:**
The environment was stress-tested with 100 random-action episodes in 
Week 1 to confirm stability (0 crashes, no invalid states).
