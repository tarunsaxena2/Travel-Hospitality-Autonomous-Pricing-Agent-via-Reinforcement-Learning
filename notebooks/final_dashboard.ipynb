## 1. Project Summary

### Goal
Train a Deep Q-Network (DQN) agent to autonomously set dynamic prices 
for travel/hospitality bookings, outperforming heuristic baselines 
(fixed, time-based discount, demand-based) in mean episodic revenue 
across 1,000 simulated seasons.

### MDP Formulation
- **State**: [remaining_inventory, days_until_departure]
- **Action**: Discrete price level bins (10 levels)
- **Reward**: price_level × units_sold
- **Horizon**: Episode ends when inventory = 0 or days_until_departure = 0

## 2. Environment Design

### Custom Gymnasium Environment: `PricingEnv`

A custom `gym.Env` subclass was built to simulate the dynamic pricing 
problem for travel/hospitality bookings.

**State Space (Observation):**
- `remaining_inventory`: units left to sell (0 to 100)
- `days_until_departure`: days left in the booking season (0 to 30)

**Action Space:**
- Discrete(10) — 10 price level bins, normalized from 0 (lowest) to 1 (highest)

**Reward Function:**

**Demand Model:**
A stochastic demand function using a logistic curve determines how many 
units sell at each step. Purchase probability:
- Decreases as price increases
- Increases as the deadline approaches (urgency effect)

**Episode Termination:**
An episode ends when either:
- `remaining_inventory` reaches 0 (sold out), or
- `days_until_departure` reaches 0 (season ended)

**Validation:**
The environment was stress-tested with 100 random-action episodes in 
Week 1 to confirm stability (0 crashes, no invalid states).
